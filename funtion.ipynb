## Q. Create a funtion to print all the palindrom numbers present between a given range.

```python
def isPalindrom(x,y):
    for i in range(x,y):
        digit = 0
        temp = i
        while i > 0:
            d = i % 10
            digit = (digit * 10) + d
            i = i// 10
        if temp == digit:
            print(temp,end=" ")
```

### Q. Create a function to find all the perfect numbers present between the specificed range given by the user.
<pre>
    When a number is equal to the sum of all its positive divisors except itself, it is said to be a perfect number.
    * Ex: 6
    sum of divisors is 1 + 2 + 3 = 6.
    Hence, 6 is a perfect number.
</pre>

```python
def isPerfect(st,end):
    for i in range(st,end):
        sum = 0
        for j in range(1,i):
            if i % j == 0:
                sum = sum + j
        if i == sum:
            print(i,end=" ")
isPerfect(int(input("Enter start number : ")),int(input("Enter end nuber : ")))
```

### Q Create a function isArmstrong and use the function to find all the armstrong numbers present between 100~2000.

```python
def isArmstrong(start,end):
    for i in range(start,end+1):
        x = len(str(i))
        sum = 0
        temp = i
        while i > 0:
            d = i % 10
            sum = sum + (d**x)
            i = i//10
        if temp == sum:
            print(temp, end = " ")
isArmstrong(int(input("Enter a start number : ")),int(input("Enter a start number : ")))
```
