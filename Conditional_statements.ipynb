# Conditional Statements/ Decision Making statement / Controls statement : 

<b>Conditional statements is also called as decision making statements or control statements. : </b>
<b>Defination : This statements are used to make some decision based on some condition according to which the control flow in a program.<b>

<b>Types of conditional statement</b>
1. Single if/simple if statments
2. If-Else statements
3. Nested if statements
4. if - elif - else statements
   

## 1. Single if : only check the truthness of the program and return an output.
<pre> Syntax : 
        if condition: 
            statements
            ......
</pre>

```python
age = int(input("Enter you age : "))
if age >= 18:
    print("You are eligible for vote.")
    print("🎉 Congratulations")
```

## 2. if-else statements :  
<b>This statements are used to check the truthness as well as the falsiness of a program and return the answer accordingly. <b>
<pre> 
    <u>syntax:</u> 
        if condition:
            statements
            ......
        else:
            statements
            .......
</pre>

```python
age = int(input("Enter you age : "))
if age >= 18:
    print("You are eligible for vote.")
    print("🎉 Congratulations")
else:
    print("So Sorry 😔 you are not eligible.")
    print(f"Waiting for {18-age} years.")
```

## 3. Nested if statements : 
<b>This helps to check mulitple conditions at a time.</b>
<pre> Syntax : 
            if condition:
                statements
                .........
                if condition:
                    statments
                    .........
                    if condition
                        statments
                        ......
            </pre>

## 4. if elif else: 
<b>It helps to check multiple conditions as options.</b>
<pre>
    <u>Syntax</u> : 
        if condition:
            statements
            .......
        elif condition:
            statments
            ......
        else:
            statements
            ......
</pre>

```python
enter_your_mark = float(input("Enter your mark : "))
if enter_your_mark >= 90:
    print("O Grade.")
elif enter_your_mark >= 80:
    print("A+ Grade.")
elif enter_your_mark >= 70:
    print("A Grade")
elif enter_your_mark >= 60:
    print("B+ Grade.")
elif enter_your_mark >= 50:
    print("B Grade.")
elif enter_your_mark >= 40:
    print("C Grade.")
else:
    print("Fail.")
```
