## Pipeline :
<pre>
    A pipeline is a tool in machine learning that connects multiple steps together and executes them in the correct order.
</pre>

```python
from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import MultinomialNB
```

```python
model = Pipeline([('vector', TfidfVectorizer()), 
                 ('tfidf', TfidfTransformer()),
                 ('algo', MultinomialNB())])
```

```python
model.fit(news_train.data, news_train.target)
```

```python
def predict_category(s,train=news_train, model=model):
    pred = model.predict([s])
    return train.target_names[pred[0]]
```

```python
predict_category('Arjun Murmu is a collage student.')
```

```python
predict_category('Draupadi Murmu is our president of india.')
```
