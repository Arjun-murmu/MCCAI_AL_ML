### Stemming : 
<pre>
    Stemming is the process of remove suffixes from a word to reduce ot to its base or root from called stemming.
    - it finds the nearest root word by spelling 
    Playing --> play
    Happier --> Happi
</pre>

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
ps.stem('cars')
```

```python
ps.stem('study')
```

```python
ps.stem('happiness')
```

### to find puntuation make present in the string.

```python
txt3 = 'Wow !!! this is so pretty.....?'
txt3
```

```python
import string
for c in txt3:
    if c in string.punctuation:
        print(c,end=" ")
```

```python
# using lambda function
punct = [c for c in txt3 if c in string.punctuation]
print(punct)
```

```python
import string
for c in txt3:
    if c not in string.punctuation:
        print(c,end='')
```

```python
# using lambda function
punct = [c for c in txt3 if c not in string.punctuation]
print(punct)
```

## Vectorizer :

```python
# count vectorizer
from sklearn.feature_extraction.text import CountVectorizer
txt4 = ['I Love Singing and I Love Dancing']
cv = CountVectorizer()
x = cv.fit_transform(txt4)
```

```python
print(x)
```

```python
cv.vocabulary_
```

```python
print(x.toarray())
```

```python
# for multiple sentence
txt5 = ['I Love Singing and I Love Dancing', 'I Love Ice-Creame']
y = cv.fit_transform(txt5)
```

```python
print(y)
```

```python
cv.vocabulary_
```

```python
print(y.toarray())
```

## TF-IDF Vectorizer
