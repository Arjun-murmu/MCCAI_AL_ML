# Slicing

## It is technique used to extract a specific portion of elements from a sequence.

```python
food = "Ice Cream"
food
```

```python
# forward slicing
# syntax = var[start:stop+1:step]
print(food[0:9:2])
print(food[::1])
print(food[:3:])
print(food[-1::])
print(food[-1:-6:])
print(food[-1:-10:-1])
print(food[-1:3:-1])
```

```python
z = 'AEROPLANE'
# print(z[0:9:2])
print(z[-5:-9:-1])
print(z[4:0:-1])
```

```python
print(z[-2:-7:-1])
print(z[7:2:-1])
```

```python
# Default Indexing
# z[2:6:1]
z[2:6:]
```

```python
z[:9]
```

```python
z[::1]
```

```python
z[::-1]
```

```python
z[:-5:-1]
```

```python
p = 'AEROSPACE'
print(p[-2:-6:-1])
print(p[::2])
print(p[-7:-10:-1])
print(p[6::])
print(p[::-1])
```
