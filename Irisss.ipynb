### 3.ENCODING

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
iris.Flower_Type=le.fit_transform(iris.Flower_Type)
```

```python
iris['Flower_Type'].unique()
```

```python
c=iris.corr()
plt.figure(figsize=(20,15))
sns.heatmap(c,annot=True)
plt.show()
```

### 4.IP/OP CREATIONS

```python
ip=iris.drop('Flower_Type',axis=1)
ip
```

```python
op=iris.Flower_Type
op
```

### 5.TRAIN-TEST SPLIT

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(ip,op,test_size=0.2)
```

```python
print(x_train.shape)
print(x_test.shape)
```

### 6.STANDARD SCALER

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

### 7.MACHINE LEARNING MODESL

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train,y_train)
```

### 8.PREDICTION

```python
pred=lr.predict(x_test)
pred
```

### 9.ACCURACY

```python
from sklearn.metrics import accuracy_score,recall_score,f1_score,precision_score
acc=accuracy_score(y_test,pred)
recall=recall_score(y_test,pred,average='macro')
f1=f1_score(y_test,pred,average='macro')
precision=precision_score(y_test,pred,average='macro')
print("Accuracy Score:",acc)
print("Recall Score:",recall)
print("F1 Score:",f1)
print("Precision Score:",precision)
```

```python
#confusion matrix
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
cm=confusion_matrix(y_test,pred)
cm
```

```python
cmd=ConfusionMatrixDisplay(cm)
cmd.plot()
plt.show()
```

### KNN(K-Nearest Neighbors)

```python
from sklearn.neighbors import KNeighborsClassifier
Knn=KNeighborsClassifier(n_neighbors=5)
Knn.fit(x_train,y_train)
```

```python
pred2=Knn.predict(x_test)
pred2
```

```python
from sklearn.metrics import accuracy_score,recall_score,f1_score,precision_score
acc=accuracy_score(y_test,pred2)
recall=recall_score(y_test,pred2,average='macro')
f1=f1_score(y_test,pred2,average='macro')
precision=precision_score(y_test,pred2,average='macro')
print("Accuracy Score:",acc)
print("Recall Score:",recall)
print("F1 Score:",f1)
print("Precision Score:",precision)
```

```python
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
cm=confusion_matrix(y_test,pred2)
cm
```

```python
cmd=ConfusionMatrixDisplay(cm)
cmd.plot()
plt.show()
```
