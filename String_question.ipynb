#  1. WAPP to take input your full name and should only rreturn the surname.

```python
name = input("Enter Your full name : ")
namelist = name.split()
print(namelist[-1])

# Why is a list used?
# Because split() helps convert a single string into a list of multiple strings. When I use split(), the name is converted into a list like ['Arjun', 'Murmu']. 
#Then, I store this list in the variable 'namelist'. Each word has a specific index number.
# For example:
# Arjun has index number 0, and Murmu has index number 1.
# In backward indexing, Arjun is -2, and Murmu is -1.
#
# We know that the surname usually comes at the end of a full name.
# Therefore, using namelist[-1] returns the last word of the name,
# which is usually the surname.
#
# Slicing can also be used, but it may not work correctly
# when the user's input contains a different number of words.
```

```python
s = "A man, a plan, a canal: Panama"
# s = s.lower()
p = ""
for i in s:
   if i.isalnum():
       p += i
print(p.lower())
```

```python
clean = ""
for i in s:
    if i.isalnum():
        clean += i
s = clean.lower()
clean = clean.lower()
print(clean)
rev_s = clean[::-1]
print("rev : ",rev_s)
print("s : ",s)
if s == rev_s:
    print("True")
else:
    print("False")
        
```
