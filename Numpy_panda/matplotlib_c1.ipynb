# Matplotlib :

```python
# import matplotlib 
import matplotlib.pyplot as plt
```

## Line chart
<b> a line chart is a graph that is used to connected points to show how data changes or move overtime </b>

```python
x = [1,2,5,7,9,11,13]
y = [2,4,6,8,10,12,14]
plt.plot(x,y)
```

```python
p = [2000,2001,2002,2003]
q = [40,50,45,80]
plt.plot(p,q)
plt.show()
```

```python
p = [2000,2001,2002,2003]
q = [40,50,45,80]
plt.plot(p,q,c='yellow')
plt.show()
```

```python
p = [2000,2001,2002,2003]
q = [40,50,45,80]
plt.plot(p,q,c='blue')
plt.show()
```

```python
p = [2000,2001,2002,2003]
q = [40,50,45,80]
plt.plot(p,q,'red')
plt.title('sales line plot')
plt.show()
```

```python
x = [1,2,5,7,9,11,13]
y = [2,4,6,8,10,12,14]
plt.plot(x,y,'blue')
plt.title('Line chart')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.show()
```

```python
x = [1,2,5,7,9,11,13]
y = [2,4,6,8,10,12,14]
plt.plot(x,y,'blue',marker='.')
plt.title('Line chart')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.show()
```

```python
x = [1,2,5,7,9,11,13]
y = [2,4,6,8,10,12,14]
plt.plot(x,y,'blue',marker='o',linestyle='--')
plt.title('Line chart')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.show()
```

```python
x = [1,2,5,7,9,11,13]
y = [2,4,6,8,10,12,14]
plt.plot(x,y,'blue',marker='o',linestyle='--')
plt.title('Line chart')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.grid()
plt.show()
```

```python
x = [1,2,5,7,9,11,13]
y = [2,4,6,8,10,12,14]
plt.plot(x,y,'blue',marker='o',linestyle='--')
plt.title('Line chart')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.grid()
plt.show()
```

## bar plot:
<b> it represent data using rectangular bars, where the height of each bar shows the values of each category.</b>


```python
m = ['Team A','Team B','Team C','Team D']
n = [500,450,670, 250]
plt.bar(m,n)
plt.title('Bar plot')
plt.xlabel('Teams')
plt.ylabel('Number of project')
plt.show()
```

```python
m = ['Team A','Team B','Team C','Team D']
n = [500,450,670, 250]
plt.bar(m,n,color=['y','pink','g','r'])
plt.title('Bar plot',color='blue')
plt.xlabel('Teams',color='purple')
plt.ylabel('Number of project',color='purple')
plt.show()
```

```python
# multiple bar plot
p = ['Team A','Team B','Team C','Team D']
q1 = [500,450,670, 250]
plt.bar(p,q1)
p = ['Team A','Team B','Team C','Team D']
q2 = [550,480,700, 300]
plt.bar(p,q2,width=0.5)
plt.title('Bar plot',color='blue')
plt.xlabel('Teams',color='purple')
plt.ylabel('Number of project',color='purple')
plt.show()
```

## Histogram : 
<b> a histogram is a graph that shows the distribution of numerical data by grouping values into the ranges.</b>

```python
age=[23,67,45,89,74,90,58,69,23]
interval=[20,30,40,50,60,70,80,90]
plt.title('Histogram Graph')
plt.xlabel('Interval')
plt.ylabel('age')
plt.hist(age,interval)
plt.show()
```

# scatter plot:
<b> scatter plot is used to show the relationship between two numerical variables. Each points reprsent one observation.</b>

```python
study_hrs = [3,4,2,5,1]
result = [70,80,50,95,10]
plt.scatter(study_hrs, result,marker='^',s=200)
plt.show()
```

# pie plot/pie chart : 
<b> pie plot used to display the data as a prepration of any whole values</b>

```python
activites = ['studying','food','romaing','playing','sleeping']
slice = [3,2,8,3,8]
plt.pie(slice,labels=activites)
plt.show()
```

```python
activites = ['studying','food','romaing','playing','sleeping']
slice = [3,2,8,3,8]
col = ['pink','violet','gold','purple','lightblue']
plt.pie(slice,labels=activites,colors=col)
plt.show()
```

```python
activites = ['studying','food','romaing','playing','sleeping']
slice = [3,2,8,3,8]
col = ['pink','violet','gold','purple','lightblue']
plt.pie(slice,labels=activites,colors=col,autopct='%1.2f%%')
plt.show()
```

```python
activites = ['studying','food','romaing','playing','sleeping']
slice = [3,2,8,3,8]
col = ['pink','violet','gold','purple','lightblue']
plt.pie(slice,labels=activites,colors=col,autopct='%1.2f%%',shadow=True)
plt.show()
```

```python
activites = ['studying','food','romaing','playing','sleeping']
slice = [3,2,8,3,8]
col = ['pink','violet','gold','purple','lightblue']
plt.pie(slice,labels=activites,colors=col,autopct='%1.2f%%',explode=(0,0,0.2,0,0.2))
plt.show()
```
