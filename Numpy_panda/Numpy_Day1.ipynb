*
#### NumPy Logarithm Functions

NumPy provides the following logarithm functions:

* `np.log(x)` → Calculates the natural logarithm (base `e`).
* `np.log2(x)` → Calculates the logarithm with base `2`.
* `np.log10(x)` → Calculates the logarithm with base `10`.

<pre>
Example:

np.log(10)
np.log2(8)
np.log10(100)
</pre>
For logarithms with other bases, use the change-of-base formula:
<pre>
np.log(x) / np.log(base)
Example:
log4 of 16 is 2 then used this type : 
    np.log(16) / np.log(4)    # 2.0
</pre>
NumPy does not provide separate functions like `np.log3()`, `np.log4()`, `np.log5()`, etc.


```python
print(np.log(10))
print(np.log2(8))
print(np.log10(1000))
```

```python
print("log4(16) = ",np.log(16) / np.log(4))
print("log3(81) = ",np.log(81)/np.log(3))
```

```python
l1 = np.log(16)
l2 = np.log(4)
print(f"log(16) = {l1} and log(4) = {l2} and division value is : {l1/l2}")
```
