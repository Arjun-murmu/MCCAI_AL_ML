<pre>

```python
# importing pandas
import pandas as pd
```

```python
# Version check of pandas
print(pd.__version__)
```

```python
# Creating a series : 
s = pd.Series(['a','b','c','d'])
print(s)
print("Types of s : ", type(s))
print("Address of s : ",id(s))
```

```python
# Assign index position 
s1 = pd.Series(['a','b','c','d'], index=[10,20,30,40])
print("index  value \n",s1)
```

```python
s2 = pd.Series(['arjun','bhai','chua','don'], index=['a','b','c','d','e'])
print("index  value \n",s2)
```

```python
s2 = pd.Series(['arjun','bhai','chua','don','eye','future'], index=[10,20,20,30,40,50])
print("index  value \n",s2)
print("Indexing s2[20] : \n",s2[20])
```

```python
# Slicing of pandas
print("s2[0:3:1] : \n", s2[0:3:1])
print("s2[::-1] : \n", s2[::-1])
```

```python
s2[10:30:1]
```

```python
# .loc : look at the INDEX
# if the change the index position then used the .loc for access as per the user indexing position
s2.loc[10:40:1]
```

```python
print(s2[20])
```

```python
# Tuple to series  
t = pd.Series((10,20,30,40))
print(t)
print(t[2])
print(t[0:3:1])
```

```python
print(t[-2])
# we can't use -ve indexing on a Series.
```

```python
# for -ve indexing we can use iloc
print(t.iloc[-1])
print(t.iloc[-2])
```

```python
# list to Series
l = pd.Series([100,200,300,400])
print(l)
print(l.iloc[-1])
```

```python
# set to Series
se = pd.Series({10,20,30,40,50})
se
```

```python
# Dictionary to series :
D = pd.Series({'a' : 1,'b' : 2, 'c' : 3, 'd' : 4})
print(D)
```

```python
E = pd.Series({'a' : 1,'b' : 2, 'c' : 3, 'd' : 4}, index={'e','b','f','d'})
print(E)
```

```python
print(E[2])
print(E['b'])
print(E['d'])
```

```python
print(E[0:3:1])
print(E[::1])
print(E[::-1])
```

```python
print(E['a':'d':1])
```

### DataFrame:
<b> </b>

```python
# creating a DataFrame : 
DF = {'Name' : ['Arjun','Site','Hari','Subha','Bhagya'],
    'Roll': [21,31,41,51,61],
     'Skills': ['Cooking','Travel','Sleeping','Study','Foodie']
     }
print(DF)
```

```python
data = pd.DataFrame(DF)
data
```

```python
# indexing
print("4th index name : ",data.Name[3])
print("3th index name : ",data.Name[2])
print("4th index Skills : ",data.Skills[3])
print("3th index Skills : ",data.Skills[2])
print("4th index Roll : ",data.Roll[3])
print("3th index Roll : ",data.Roll[2])
```

```python
# Slicing
print(data[1:3:1])
print("\n")
print(data[0:3:2])
```

```python
print(data[4:5:1])
```

```python
# iloc(integer Location) : used for slicing on a DataFrame
# iloc[row,column]
data.iloc[3,::]
```

```python
# column
data.iloc[::,2]
```

```python
data.iloc[::-2,::]
```

```python
data.iloc[::-1,::-1]
```

```python
print(data.Name)
print(data.Skills)
print(data.Roll)
```

```python
data
```

```python
# head : head it repesent the top rows (bydefault 5)
data.head(3)
```

```python
# tail : it represent the bottom rows (bydefault 5)
print(data.tail())
data.tail(2)
```

```python
# sample : represent random row/sample (bydefault one)
print("1st Run : \n",data.sample())
print("2nd Run : \n",data.sample())
print("3rd Run : \n",data.sample())
```

```python
data.sample(3)
```
