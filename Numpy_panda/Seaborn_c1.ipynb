```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# to hide  warnings 
import warnings 
warnings.filterwarnings('ignore')
```

```python
# distplot
sns.distplot(iris.sepal_length)
plt.show()
```

```python
# displot() : show data distribution using bars
sns.displot(iris.sepal_length)
plt.show()
```

```python
# histplot() : show data count using bars.
sns.histplot(iris.sepal_width)
plt.show()
```

```python
#kdeplot : show data distribution using curvature lines
sns.kdeplot(iris.sepal_length)
plt.show()
```

```python
#kdeplot : show data distribution using curvature lines
sns.kdeplot(iris.sepal_length,fill=True,color='yellow')
plt.show()
```

```python
iris['sepal_length'].describe()
```

```python
# boxplot() : used to display the statical values for any numeric/continuous variable.
sns.boxplot(y=iris.sepal_length)
plt.grid()
plt.show()
```

```python
sns.boxplot(x = iris.Flower_Type, 
            y = iris.sepal_length,palette='plasma')
plt.grid()
plt.show()
```

```python
# violin plot : 
sns.violinplot(x = iris.Flower_Type,
             y = iris.sepal_length)
plt.grid()
plt.show()
```

```python
# violin plot : 
sns.violinplot(x = iris.Flower_Type,
             y = iris.sepal_length,palette='Greens')
plt.grid()
plt.show()
```

```python
# violin plot : 
sns.violinplot(x = iris.Flower_Type,
             y = iris.sepal_width, palette='Greens')
plt.grid()
plt.show()
```

```python
# Boxen plot : used to represent the statistical data using quartiles.
sns.boxenplot(iris.petal_width)
plt.grid()
plt.show()
```

```python
print("50% : ",np.percentile(iris.petal_width,50.0))
print("25% : ",np.percentile(iris.petal_width,25.0))
print("75% : ",np.percentile(iris.petal_width,75.0))
```

```python
# swarm plot : used data point which are adjusted along the categorical axis to prevent overlap.
# every indivisual observation is visible.
sns.swarmplot(x = iris.Flower_Type, y = iris.petal_width)
plt.grid()
plt.show()
```

```python
sns.swarmplot(data=iris)
plt.show()
```

```python
# barplot() : It used to compare values between different categories.
sns.barplot(x = 'Flower_Type', y = 'petal_length', data = iris)
plt.show()
```

```python
print(iris.groupby('Flower_Type')['petal_length'].mean())
```

```python
# scatter plot :it used To show the relationship between numerical variable.
sns.scatterplot(x = 'sepal_length',y = 'petal_length', data = iris)
plt.show()
```
