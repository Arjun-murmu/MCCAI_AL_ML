```python
ip = pima.drop('Outcome', axis=1)
ip
```

```python
op = pima.Outcome
op
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(ip,op,train_size=0.8)
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
x_train
```

```python
x_test
```

```python
#Machine Learning Models LoLogisticRegression
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train,y_train)
```

```python
pred = lr.predict(x_test)
pred
```

```python
from sklearn.metrics import accuracy_score,recall_score,f1_score,precision_score
acc=accuracy_score(y_test,pred)
recall=recall_score(y_test,pred)
f1=f1_score(y_test,pred)
precision=precision_score(y_test,pred)
print("Accuracy Score:",acc)
print("Recall Score:",recall)
print("F1 Score:",f1)
print("Precision Score:",precision)
```

```python
# Confusion matrix:
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
cm=confusion_matrix(y_test,pred)
cm
```

```python
cmd=ConfusionMatrixDisplay(cm)
cmd.plot()
plt.show()
```

```python
#Machine Learning Models KNN(k_Nearest_Neighbors):
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=10)
knn.fit(x_train,y_train)
```

```python
pred2 = knn.predict(x_test)
pred2
```

```python
from sklearn.metrics import accuracy_score,recall_score,f1_score,precision_score
acc=accuracy_score(y_test,pred2)
recall=recall_score(y_test,pred2)
f1=f1_score(y_test,pred2)
precision=precision_score(y_test,pred2)
print("Accuracy Score:",acc)
print("Recall Score:",recall)
print("F1 Score:",f1)
print("Precision Score:",precision)
```

```python
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
cm=confusion_matrix(y_test,pred2)
cm
```

```python
cmd=ConfusionMatrixDisplay(cm)
cmd.plot()
plt.show()
```
