### SEABORN:
<PRE>Seaborn is a python library used for data visualization.
it is built on top of metplotlib and provides many attractive statistical graphs.</PRE>

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#to hide warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
#distplot
sns.distplot(iris.sepal_length)
plt.show()
```

```python
#displot():show data distribution using bars 
sns.displot(iris.sepal_length)
plt.show()
```

```python
#histplot():shows data count using bars.
sns.histplot(iris.sepal_width,color='g')
plt.show()
```

```python
#kdeplot():show data distribution using curvature lines.
sns.kdeplot(iris.sepal_length,fill=True,color='Pink')
plt.show()
```

```python
#boxplot:used to display the statistical values for any numeric/continuous variable.
sns.boxplot(y=iris.sepal_length)
plt.grid()
plt.show()
```

```python
iris['sepal_length'].describe()
```

```python
sns.boxplot(x=iris.Flower_type,
           y=iris.sepal_length,palette='plasma')
plt.grid()
plt.show()
```

```python
#violin plot:
sns.violinplot(x=iris.Flower_type
               ,y=iris.sepal_length,palette='Greens')
plt.grid()
plt.show()
```

```python
#Boxen plot:used to represent the statistical data using quartiles.
sns.boxenplot(iris.petal_width)
plt.grid()
plt.show()
```

```python
np.percentile(iris.petal_width,50.0)
```

```python
#swarm plot:used data points which are adjusted along the categorical axis to prevent overlap.every indivisual observation is visible.
sns.swarmplot(x=iris.Flower_type,y=iris.petal_width)
plt.grid()
plt.show()
```

```python
sns.swarmplot(data=iris)
plt.show()
```

```python
#Bar plot():used to compare values between different categories.
sns.barplot(x='Flower_type',y='petal_length',data=iris)
plt.show()
```

```python
iris.groupby('Flower_type')['petal_length'].mean()
```

```python
#Scatter plot: It is used to show the relationship between two numeric variables.
sns.scatterplot(x='sepal_length',y='petal_length',data=iris,hue='Flower_type')
plt.show()
```

```python
iris[['sepal_length','petal_length']]
```

### Joint plot:
<pre>It is used to understand the relationship between two numerical variable and their individual distribution at the same time.</pre> 

```python
sns.jointplot(x=iris.sepal_length,y=iris.petal_length)
plt.show()
```

```python
sns.jointplot(x=iris.sepal_length,y=iris.petal_length,data=iris,hue='Flower_type')
plt.show()
```

### Count plot: 
<pre>It is used to count how many times each category appears in a dataset.</pre>

```python
sns.countplot(x='Flower_type',data=iris)
plt.show()
```

```python
sns.countplot(iris.Flower_type)
plt.show()
```

### Pair plot:
<pre>It is used to see the relationship between multiple numerical columns at the same time.</pre>`

```python
sns.pairplot(iris,hue='Flower_type')
plt.show()
```
