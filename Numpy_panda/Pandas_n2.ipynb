### to replace null values

```python
DF2 = pd.DataFrame({'Days' : [2,5,np.nan,8,18],
    'Place': ['Puri','Manali','Goa','Puri',np.nan],
     'Visitor': [5590,5000,4895,2500,3400],
       'Mode' : ['Bus','Train','Bus','Bike','Cycle']
     })
DF2
```

```python
DF2.shape
```

```python
DF2.info()
```

```python
DF2.isnull()
```

```python
DF2.isnull().sum()
```

```python
DF2.dtypes
```

```python
# fillna :  if null value then we have to use fillna
# replace : if special charactet then we have to use replace
```

```python
DF2.describe(include='all')
```

```python
# fillna-used to replace the null values
# in case of numeric column we replace 'nan' with mean/Avg
DF2.Days = DF2.Days.fillna(8.250000)
```

```python
DF2
```

```python
DF2.Place = DF2.Place.fillna('Puri')
```

```python
DF2
```

```python
DF2.isnull().sum()
```

```python
DF3 = pd.DataFrame({'Days' : [2,5,7,8,18],
    'Place': ['Puri','Manali','Goa','Puri','?'],
     'Visitor': [5590,5000,4895,"?",3400],
       'Mode' : ['Bus','Train','Bus','Bike','Cycle']
     })
DF3
```

```python
DF3.shape
```

```python
DF3.info()
```

```python
DF3.dtypes
```

```python
DF3.isnull().sum()
```

```python
# unique
DF3.Place.unique()
```

```python
DF3.Visitor.unique()
```

```python
DF3.describe(include='all')
```

```python
DF3.Place = DF3.Place.replace('?','Puri')
```

```python
DF3
```

```python
# in case of "?" in numeric column
# replace "?" with np.nan
DF3.Visitor = DF3.Visitor.replace('?',np.nan)
```
