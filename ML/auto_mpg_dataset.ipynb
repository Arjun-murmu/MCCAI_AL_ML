# Linear Regression

```python
ip = auto2.drop(['mpg'], axis=1)
ip
```

```python
op = auto2['mpg']
```

```python
# train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(ip,op,test_size=0.3)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.fit_transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
alg = LinearRegression()
alg.fit(x_train,y_train)
```

```python
yp = alg.predict(x_test)
```

```python
yp
```

```python
df = pd.DataFrame({'y' : list(y_test), 'predication' : yp})
```

```python
yp.shape
```

```python
df
```

```python
x = np.array([10,307.0,130.0, 3504.0, 12.0, 70,1]).reshape(1,-1)
sc.fit(x)
x = sc.transform(x)
alg.predict(x)
```

```python
# accuracy
from sklearn.metrics import mean_squared_error, r2_score
mse = mean_squared_error(y_test,yp)
r2 = r2_score(y_test,yp)
print("Mean Square Error : ",mse)
print("R2 Score : ", r2)
```

```python
print('m', alg.coef_)
print('c', alg.intercept_)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2)
x_poly = poly.fit_transform(x_train)
pol_reg = LinearRegression()

pol_reg.fit(x_poly,y_train)
x1 = poly.fit_transform(x_test)
yp2 = pol_reg.predict(x1)
```

```python
# accuracy
from sklearn.metrics import mean_squared_error, r2_score
mse = mean_squared_error(y_test,yp2)
r2 = r2_score(y_test,yp2)
print("Mean Square Error : ",mse)
print("R2 Score : ", r2)
```

```python
# Ridge Regression
from sklearn.linear_model import Ridge
R = Ridge(alpha=0.9)
R.fit(x_train,y_train)
yp1 = R.predict(x_test)

from sklearn.metrics import mean_squared_error, r2_score
mse1 = mean_squared_error(y_test,yp1)
rr2 = r2_score(y_test,yp1)
print("Mean Square Error : ",mse1)
print("R2 Score : ", rr2)
```

```python
# Lasso Regression
from sklearn.linear_model import Lasso
L = Lasso(alpha=0.09)
L.fit(x_train,y_train)
yp3 = L.predict(x_test)

from sklearn.metrics import mean_squared_error, r2_score
mse2 = mean_squared_error(y_test,yp3)
r22 = r2_score(y_test,yp3)
print("Mean Square Error : ",mse2)
print("R2 Score : ", r22)
```
