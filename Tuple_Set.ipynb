# Tuple Data Type & Attributes :- 

## Defination : A tuple is a immutable (Unchangeable) sequence of elements present inside a (). 
<pre>1. Tuples are ordered sequence. 
2. Tuple allowes duplicate values.
</pre>

```python
# Homogenous tuple : Elemests are having similar data types.
tup1 =  (12,12,40,90,60,60,80)
tup1
```

```python
# Hetrogenous Tuple : elements having different datatypes.
tup2 = (12,True,'string',78+9j)
tup2
```

```python
# Tuple is the default datatype among sequential data types in python
tup3 = 12,True,'string',78+9j
type(tup3)
```

# Set data types and Attributes

## A set is a mutable(changeable) collection of individual items present inside a {,,,,,}
<pre>1. set are unordered sequence.
2.It can be homogenous or hetrogenous.
3. set only allows to store unique collection of values.
4. set allows all mathematical operations. </pre>

```python
# Homogenous set : having similar datatype.
s1 = {12,24,56,78,90,12}
s1
```

```python
type(s1)
```

```python
# heterogenous set : having difference datatypes.
s2 = {12,12,'name',12.5,9+9j,False}
s2
```

```python
type(s2)
```

```python
# set are unordered - doesnot have fixed index positions.
s2
```

```python
# Set is unorder so you can't proform indexing and slicing.
```

```python
s2[1]
```

```python
s2[1:4:1]
```

```python
# set does not allows Duplicate values
s3 = {12,45,45,67,78,98,67}
s3
```

## set Attributes :-

```python
s4 = {12,12,'name',12.5,9+9j,False}
```

### 1. add() : it allows to add new element to the set.


```python
s4.add('ramu')
s4.add(123)
s4
```

```python
# Multiple value can't possible to add at time
s4.add('cow','ox')
```

### 2. Update() : allows to add multiple values to the set.

```python
s5 = {12,12,'name',12.5,9+9j,False}
s5.update(('cow','monkey','dog',55,66))
```

```python
s5
```

```python
s6 = {12,12,'name',12.5,9+9j,False}
s6.update(['cow','monkey','dog',55,66])
s6
```

```python
s7 = {12,12,'name',12.5,9+9j,False}
s7.update({'cow','monkey','dog',55,66})
s7
```

### 3. remove() : It deletes a specific elements from a sequence.

```python
s8 = {12,12,'name',12.5,9+9j,False}
s8.update({'cow','monkey','dog',55,66})
s8
```

```python
s8.remove(12)
s8
```

### 4. discard() : it delete a specific element from a sequence.

```python
s9 = {12,12,'name',12.5,9+9j,False}
s9.update({'cow','monkey','dog',55,66})
s9
```

```python
s9.discard(55)
s9
```

### 5. pop() : Delete any random value from the set.

```python
s10 = {12,12,'name',12.5,9+9j,False}
s10.update({'cow','monkey','dog',55,66})
s10
```

```python
s10.pop()
s10
```

```python
s10.pop()
```

```python
sa13 = {12,12,'name',12.5,9+9j,False}
sb13 = {'cow','monkey','dog',55,66,12,'name'}
```

```python
# union() : to return all the elements present in different sets.
sa13.union(sb13)
```

```python
# intersection() : return only the common values
sa13.intersection(sb13)
```

```python
# difference() : return the unique value after excluding the common values.
sa1 = {1,2,3,4,5}
sb1 = {'a','b',1,2}
sa1.difference(sb1)
```

```python
sb1.difference(sa1)
```

```python
# symmetric_difference() = it combinely returns the unique values present in both set after deleting the common value
sa2 = {1,2,3,4,5}
sb2 = {'a','b',1,2}
sa2.symmetric_difference(sb2)
```

```python
# issuperset() : return true if one set contains aother set completely along with same extra values.
sa3 = {1,2,3,4,5}
sb3 = {1,2}
sa3.issuperset(sb3)
```

```python
# issubset() : return true if one set contains aother set completely common to another set.
sa4 = {1,2,3,4,5}
sb4 = {1,2}
sa4.issubset(sb4)
```

```python
sb4.issubset(sa4)
```

```python
# isdisjoint() : returns true if the set doesnot contains any common/joining values.
sa5 = {1,2,3,4,5}
sb5 = {1,2}
sa5.isdisjoint(sb5)
```

```python
sb5.isdisjoint(sa5)
```

```python
sa5 = {1,2,3,4,5}
sb6 = {'a','b','c','d'}
sa5.isdisjoint(sb6)
```

```python
sb6.isdisjoint(sa5)
```
