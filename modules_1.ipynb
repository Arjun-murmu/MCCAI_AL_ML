<b> Random Module : </b>
<pre>Helps to generate random numbers from a sequence, range or intervals.</pre>

```python
import random 
```

```python
# random() : Returns a randomly generated decimal value between 0 & 1
random.random()
```

```python
# uniform() : returns a randomly generated decimal value between any given range.
random.uniform(10,20)
```

```python
# randint() : return a randomly generated integer value between a given range.
random.randint(1,30)
```

```python
# randrange() : return a randomly generated integer value between a given range.
random.randrange(19,25)
```

```python
# choice() : returns a randomly chosen value from a sequence.
seq = ['apple','kutash','bileyi','mendah','badmas','chura']
random.choice(seq)
```

```python
# shuffle() : returns a jumbled up sequence.
seq = ['apple','kutash','bileyi','mendah','badmas','chura']
random.shuffle(seq)
seq
```

```python
# sample() : returns multiple randomly chosen values from a given sequence.
seq = ['apple','kutash','bileyi','mendah','badmas','chura']
random.sample(seq,4)
```

### Q. Wapp to design a game "Guess the correct number" using random module.


```python
import random
```

```python
n = random.randint(1,10)
print(n)
```

```python
n = random.randint(1,10)
while True:
    u_i = int(input("Guess the number between (1 to 10) : "))
    if u_i == n:
        print("Wow you guess number is perfect.")
        break
    elif u_i > n:
        print("Your guess number is more higher, think small number.")
    else:
        print("Your guess number is very small , think higer number.")
```

### Calender Module : deals with calendars

```python
import calendar
```

```python
# calendar() : To print any specific year.
print(calendar.calendar(2027))
```

```python
print(calendar.calendar(2050, m=5))
```

```python
print(calendar.calendar(2050, c=10))
# no of spaces row
```

```python
print(calendar.calendar(2050, m=4,c=12))
```

```python
# to print a particular month of a year.
print(calendar.month(2026,9))
```

```python
# to check a year is leap year or not : 
print(calendar.isleap(2000))
print(calendar.isleap(2012))
print(calendar.isleap(2014))
print(calendar.isleap(2026))
```

```python
# to find the total no.of leap years present in specific year range
print(calendar.leapdays(2000,2030))
```

```python
# to return the name of a month.
print(calendar.month_name[5])
```

```python
for i in range(1,13):
    print(f"{i} -> {calendar.month_name[i]}")
```

```python
for i in range(1,13):
    print(f"{i} -> {calendar.month_abbr[i]}")
```

```python
# day_name() : prints the week name. 
# 0 to 6 , where 6 -> sunday and 0-> monday
print(calendar.day_name[0])
```

```python
for i in range(7):
    print(f"{i} -> {calendar.day_name[i]}")
```

```python
# day_abbr() : prints the weeknames in shortform
for i in range(7):
    print(f"{i} -> {calendar.day_abbr[i]}")
```
