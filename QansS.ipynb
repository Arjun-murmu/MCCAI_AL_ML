# Q6. Check whether a string is palindrome

### Defination of palindrom number : A palindrome number is a number that stays the same when you flip its digits backward. It reads the same from left to right and right to left, like 121, 484, or 12321

```python
stringrpalindrom = input("Enter a number check palindrom or not  : ")

if stringrpalindrom == stringrpalindrom[::-1]:
    print("Yes this String is palindrom.")
else:
    print("This String not palindrom.")
```

# Q7. Count vowels in a string

```python
# Vowels letter like : a,e,i,o,u & A,E,I,O,U
vowel_letter = ['A','E','I','O','U','a','e','i','o','u']
enter_string = input("Enter your name check how many vowel letter : ")
vowel_let = 0
for i in range(len(enter_string)):
    # print(enter_string[i])
    if enter_string[i] in vowel_letter:
        vowel_let = vowel_let + 1
print(vowel_let)
```

    1. Multiple Inheritance + Method Resolution Order (MRO)
Create the following class hierarchy:
Person → Employee, Student → TeachingAssistant
Person: name
Employee: employee_id, salary
Student: student_id, course
TeachingAssistant: inherits from both Employee and Student.
Use constructors and super() properly. Create a TeachingAssistant object, display all details, and
determine the MRO of TeachingAssistant
</pre>

```python
class Person:
    def person(self,name):
        print("Name : ",name)
class Employee(Person):
    def emp(self, employee_id,salary):
        print(f"Employee_id : {employee_id} and salary : {salary}")
class Student:
    def stu(self, student_id, course):
        print(f"Student Id : {student_id} and course : {course}")
class  TeachingAssistant(Employee, Student):
    def display(self):
        super().person("Arjun Murmu")
        super().emp(12,23400)
        super().stu("STD001","MCCAI")
obj1 = TeachingAssistant()
obj1.display()
```
