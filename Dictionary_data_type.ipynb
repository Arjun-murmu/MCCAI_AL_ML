# TypeCasting : It is the process of changing one's datatype to another datatype explicitly.
<pre>
    Syntax : datatype(variable_name)
</pre>

```python
list_a = [56,89,234,90,56,77]
list_b = [90,56,21,96,45,89]
s1 = set(list_a)
s2 = set(list_b)
s1.symmetric_difference(s2)
print(s1,s2)
print(s1)
```

```python
print(set(list_a).symmetric_difference(set(list_b)))
```

# Dictionary Data Type and Attributes :-
That dictionary is a mutable(changeble) collection of paired elements {keys & values} present inside a {,,,}.
<pre>
    1. Dictionary does not follows any orders.
    2. It can only contains paired sequences.
    3. The keynames must be unique in the sequence.Where as values can b
    4. Each key values paired are separeted using comma.
    5. The keys are separeted from values using colon ":".
    6. In a dictionary key name can only be declared using immutable(Unchangeable) data types where a value can be declared using any of the mutable/immutable datatypes.
</pre>

## indexing and slicing : In a dictioary, we can access any specific pair using its keyname.

```python
d = {1 : 'Apple',"Mango" : 3400,True:78+8j,'a':[1,2,3]}
print(d[1])
print(d['a'])
```

```python
print(d[1:'a'])
#Dictionary can't allows slicing because it can't follow any proper sequence.
```

### Attributes : 

```python
details = {"name" : "Rajendra","id" : 101,"city" : "BBSR"}
```

```python
# 1. key() : returns only the keynames used in the dictionary.
details.keys()
```

```python
# 2. Values() : returns only the values used in the sequence.
details.values()
```

```python
# 3. items() : returns all the paired elements of the sequence.
details.items()
```

```python
# 4. get() : get return the specific values of a given keyname.
details = {"name" : "Rajendra","id" : 101,"city" : "BBSR"}
print(details.get('name'))
print(details.get('id'))
```

```python
# 5. copy() : To create a same dictionary stored in a new variable.
new = details.copy()
print(new)
```

```python
# 6. update() : update is the only attributes helps to add new pairs into the sequence.
details = {"name" : "Rajendra","id" : 101,"city" : "BBSR"}
details.update({"age" : 20})
```

```python
details.update({"age" : 20,"courese" : "AIML","Education":"Graduation"})
print(details)
```

```python
details.update({"age":32})
details
```

```python
# 7. pop() : Deletes the paired using the sepecific keyname
print(new)
new.pop("city")
print(new)
```

```python
# 8.popitem() : The popitem only deletes the last inserted pair.
new.popitem()
```

```python
# 9.clear() : deletes all pairs from the sequence.
new.clear()
new
```

```python
new_dic = details.copy()
new_dic
```

```python
# del
del new_dic["age"]
new_dic
```

```python
del new_dic
new_dic
```

```python
# 10. setdefault() : returns a value if the key exists or else it creates a key.
new_details = details.copy()
new_details
```

```python
new_details.setdefault("age",34)
new_details
```

```python
new_details.setdefault("Duration","5 month")
new_details
```
