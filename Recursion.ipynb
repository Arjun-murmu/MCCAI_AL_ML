# Recursion : 
* <b>Recursion is a process in which a function calls itself.</b>
* <b> A recursive function follows the substitution approach as well as the bottom-up approach.</b>
* <b> A recursive function, like a user-defined function, consists of three interdependent components:
* i) Function Definition
* ii) Function Declaration
* iii) Function Call
* In Python, the function definition of a recursive function mainly consists of two important parts:
* a) Base Condition
* b) Recursive Condition/Call
  </b>
  
  <pre>
      
  a)Base Statements or condition :  It is the last condition or the last value from where we get a new value a substitution in a program in order to get the program
  b) Recursive statements : This statement helps to continuously decress the value or reduce the value in order to reach the base statements.
  </pre>

  * Without a base condition, a recursive function will continue calling itself indefinitely and may result in a RecursionError.
    <pre>
    Syntax : 
    <b>def function_name(arg): 
        # Base condition 
        if condition: 
            return 
        # Recursive statement 
        function_name(modified_arg) 
    # Function call 
    function_name(value)
    </b>
</pre>

### Q1. Factorial 

```python
# Example of Recursion
def fact(n):
    if n == 0:
        return 1
    else:
        return n * fact(n-1)
fact(5)
```

### Q.8 Create a recursive function to print the Fibonacci series for given number.

```python
def fibo(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibo(n-2) + fibo(n-1)
n = int(input("Enter a number : "))
for i in range(n+1):
    print(fibo(i), end=" ")
```

### Q7. Create a function that finds the subtraction value of the product and sum of the digit of an integer.
<pre> Example : 
        ip >> 234
        prodcut >> 2*3*4 = 24
        sum = 2+3+4 = 9
        result = 24-9 = 15</pre>

```python
n = int(input("Enter a number : "))
n = str(n)
def ipds(n):
    sum = 0
    product = 1
    result = 0
    for i in n:
        product *= int(i)
        sum += int(i)
        result = product - sum
    return result
ipds(n) 
```

### Q6. Create a lambda function to find the sum of number from 1 to n
