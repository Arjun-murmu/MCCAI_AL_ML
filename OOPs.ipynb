### Q. Create a class BankAccount with instance attributes account_number, balance and methods deposit() and withdraw() to manage a bank account.

```python
class BankAccount:
    def __init__(self,account_number,balance):
        self.account_number = account_number
        self.balance = balance
    def deposit(self,amount):
        self.balance += amount
        print(f"₹{amount} Rs Deposit Successfull.")
        print(f"Total Balance : ₹{self.balance} Rs.")
    def withdraw(self,amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"₹{amount} Rs Account Deduct.")
            print(f"After Deduct Balance : ₹{self.balance} Rs.")
        else:
            print("Insufficient Balance.")
    def details(self):
        print("\n+++++++++++++++ Account Details +++++++++++++++")
        print(f"Account Number : {self.account_number}")
        print(f"Avl. Balance : ₹{self.balance} Rs .")
        print("\n+++++++++++++++++++++++++++++++++++++++++++++\n")
```

### Q. Create a class User with username and password as instance attributes having methods like Login(), Update_password() and Logout().

```python
class User:
    logged_in = False
    def __init__(self,username,password):
        self.username = username
        self.password = password
    def Login(self,username,password):
        if password == self.password and self.username == username:
            self.logged_in = True
            print("Logged in Successful.")
            print(f"Username : {self.username}\n")
        else:
            print("Invalide Credential.")
    def update_password(self,oldps,newps):
        if self.logged_in == True:
            if self.password == oldps:
                self.password = newps
                print("Password Change Successfull.")
            else:
                print("Wrong Password.")
        else:
            print("Login First.")
    def Logout(self):
        if self.logged_in == True:
            self.logged_in = False
            print("Logout successfully.")
        else:
            print("Login First")
```

# Encapsulation : 
## Date : 15.09.26
<pre>
Encapsulation in python two types : 
1. Protected Member
2. Private Member
<b>Protected Member </b>
    - A protected member a declare using a single '-' underscore before it variables name. 
    "_varname"
    - A protected member is easily access by an object of the same class.
    - It is also access by class directly.
    - It can be access by a method of the same class or subclass.
<b>Private Member</b>
    - A Private member is declare using '--' double underscore before  a variable name.
    "__varname"
    - A Private member can't be access by a object of the same class it can't be access by a class directly, or It is not access in a subclass.
    - A Private member only be access within a method of the same class.
</pre>

<pre>
<b>
- Encapsualtion is a mechanism of warpping up of all the dates into a single unit.
- It allows to Restrict the access the datas directly.
</b>
</pre>

```python
class Encapsule:
    var1 = "Normal/Public Attributes."
    _var2 = "Protected Attributes."
    __var3 = "Private Attributes."
    def display(self):
        print("Var1 : ",self.var1)
        print("Var2 : ",self._var2)
        print("Var3 : ",self.__var3)
        
```

```python
#1.  Accessing all the attributes using an object.
en = Encapsule()
print(en.var1)
print(en._var2)
print(en.__var3)
```

```python
#2.  Accessing all the attributes using a class.
print(Encapsule().var1)
print(Encapsule()._var2)
print(Encapsule().__var3)
```

```python
#3.  Accessing all the attributes using a method of the same class.
en = Encapsule()
print(en.var1)
print(en._var2)
print(en.display())
```

```python
class Loan:
    # customer_name = "Pratyasha Priyadarshani Murmu"
    # Loan_Amount = 40000
    # Time = 3
    def __init__(self,customer_name, Loan_Amount,Time):
        self.customer_name = customer_name
        self.Loan_Amount = Loan_Amount
        self.Time = Time
class HomeLoan(Loan):
    def simple_Interest(self):
        SI = (self.Loan_Amount * 7 * self.Time)/100
        print("For your Home Loan your simple interst = ",SI)
class CarLoan(Loan):
    def simple_Interest(self):
        SI = (self.Loan_Amount * 9 * self.Time)/100
        print("For your Car Loan your simple interst = ",SI)
class EducationLoan(Loan):
    def simple_Interest(self):
        SI = (self.Loan_Amount * 5 * self.Time)/100
        print("For your Education Loan your simple interst = ",SI)
```

## Polymorphism : 
<b>Polymorphism has been derive from to Greek Words Poly + Morphs where poly means : - Many and morphs means: - forms </b>
- When in a class we have same name methods but work differently for difference places it is present.
- polymorphism are of two types
- 1. Run time polymorphism
  2. compile time polymorphism
### RunTime Polymorphism :
   <b> It is also called as method over rideing 
   run time polymorphism can be achive in the conecpt of inheritance where the parent class and the child class will be having same name methods but they work differntly for each class it is present along with same or different arguments.</b>
### Compile time polymorphism : 
<b> Comiletime polymorphism also called as method overloading it can be acchived within a single class where a paticular class can have multiple same name methods but the arguments must be different to make the function work differently.</b>

### 1. Runtime Polymorphism : 

```python
# Example : 
# Clases with same name method & arguement.
class X:
    def info(self):
        print("class x info method.")
class Y(X):
    def info(self):
        print("class y info method.")
```

```python
pobj1 = Y()
pobj1.info()
```

```python
pobj1.info()
pobj1.info()
```

#### To Overcome the scenario ofo method overriding/runtime polymorphism, we must use "super()", to access the parent class's methods which are begin overridden by the child.

```python
# Example : 
# A super() method can be only implemented in a child class.
class X:
    def info(self):
        print("class x info method.")
class Y(X):
    def info(self):
        print("class y info method.")
        super().info()
```

```python
pobj2 = Y()
pobj2.info()
```
