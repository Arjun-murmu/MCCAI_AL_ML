## 1. While loop : It is used to execute a block of code/statements till a condition is True.
<pre>
    Three importance components of while loop are : 
    1. Initialization (Where to start the program)
    2. condition : (Based on this the program will be executed.)
    3. Incrementation/Decrementation : (To increase/decrease the numeric values inorder to end the loop.)
    
    Syntax :
    <b>
        initialize
        while condition:
            block of code
            ..............
            ..............
            inc/dec (Depends on the program.)
    </b>
</pre>

```python
x = 1
while x <= 10:
    print(x, end=" ")
    x += 1
```

```python
# print the values from 50 to 1
x = 50
while x >= 1:
    print(x, end=" ")
    x -= 1
```

```python
# Print the even value 100 to 200
i = 100
while i <= 200:
    print(i , end = " , ")
    i += 2
```

```python
i = 101
while i <= 200:
    if i % 2 == 0:
        print(i , end = " ")
    i += 1
```

### Q. Wapp to check whether a number is palindrom or not.
<pre>
    Palindrome number means if a number when reversed returns the same as original.
    Example :
        ip >> 2002
        op >> 2002
</pre>

```python
x = int(input("Enter a number : "))
np = x # to the store the original no as backup
new = 0
while x != 0:
    last = x % 10
    new = (new * 10) + last
    x = x//10
if np == new:
    print(f"{np} Number is palindrome.")
else:
    print(f" {np} Number is not palindrom.")
```

## For Loop :
<b>For loop is used for iterating over any sequence (That is a list, tuple, set, dictionary,string). It helps to execute of a block of code for each element present in a sequence atleast for once.</b>
<pre>
    Syntax : 
    <b> for i in sequence:
            statements
            ..........
    </b>
    Here : 
    for -> Keyword
    i -> iteratior : to fatch one element from your sequence at a time.
    in -> membership operator
    sequence -> collection/itierable
    statement > block of code
</pre>


```python
#Example
p = [10,20,30,40,(10,20,30),{89,00},'apple',{"name":"rajesh"}]
for i in p:
    print(i)
```

```python
# Iterating over a tuple:
for i in (1,3,4,3,5,0,9):
    print(i,end = " ")
```

```python
# Interting over a set:
for i in {12,34,5,6,12,'23','ar'}:
    print(i, end=" ")
    print(type(i),end = ", ")
```

```python
# Interting over a string
name = "Raju Bahul"
for i in name:
    print(i, end=" ,")
```

```python
# Interating over a dictonary:
d = {1:"Apple",2:"Orange",3:"Pomegranate"}
for elements in d:
    print(f"{elements} : {d[elements]}")
```

```python
# Interating over a dictonary:
d = {1:"Apple",2:"Orange",3:"Pomegranate"}
for elements in d:
    print(f"{elements} : {d.get(elements)}")
```

## Jump statement : 
<b> These statements are used to break the normal flow of the program by transfering the execution control to some different part of the program.</b>
<pre>
    There are three types of jump statement: 
    i) Break
    ii) Continue
    iii) Pass
</pre>

### i. Break Statement/keyword: 
<b> Break statement/keyword it used for terminate the program at the point of the we mention. </b>

```python
for i in range(1,11):
    print(i,end=" ")
```

```python
for i in range(1,11):
    if i == 7:
        break
    print(i,end=" ")
```

### ii. continue statements/keywords:
<b>It is used to skip the correct iteration by taking the control to the next iteration. </b>

```python
for i in range(1,11):
    if i == 7:
        continue
    print(i,end=" ")
```

### iii.pass Statement/keyword : 
<b>It is keyword used when a statement doe'st have it's block of code. It is also called as "empty block statement".</b>
* pass statement work if the statement contains it's block of code foe execution.

```python
for i in range(1,11):
```

```python
for i in range(1,11):
    pass
```

```python
for i in range(1,11):
    pass
    print(i,end=" ")
```

```python
for i in range(1,11):
    print(i,end=" ")
    pass
```

### Q. Wapp to check a number is armstrong or not,using a for loop.

```python
nu = input("Enter a number : ")
arm = 0
for i in nu:
    arm = arm + (int(i)**len(nu))
# print(type(nu), nu)
arm = str(arm)
if arm == nu:
    print("Armstrong number.")
else:
    print("Not Armstrong number.")
```

### Wapp to print the total numebr of digit or alphabate in any given string.
<pre>
    Example : 
        ip >> 'Abc123'
        op >> digit = 3
            alphabets = 3
</pre>

```python
string_i = input("Enter a string : ")
alpha = 0
digit = 0
s = 0
for i in string_i:
    if i.isalpha():
        alpha += 1
    elif i.isnumeric():
        digit += 1
    else:
        s += 1
        
print("Digit : ",digit)
print("Alphabets : ", alpha)
```
