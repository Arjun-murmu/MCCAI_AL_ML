# String
String are two type 
 ## 1. Singleline string
 ## 2. Multiline String

# 1. Singleline String : Character write between '...' or "... "

```python
# Example : 
a = 'I From Balasore.'
b = "Now I am in Bhubaneswar."
print(a)
print(b)
```

# 2. Multiline String : Character write between '''...........''' or """........"""

```python
# Example : 
c = ''' I From Balasore.
Now I am in Bhubaneswar. '''
d = """ Hello Everyone Good Morning.
Now i am here."""
print(c)
print(d)
```

# List DataTypes and Its Attributes :-

## Defination : A List is a mutuable collection of individual items sepated by coma ',' and endclosing within square breacket
## List are ordered sequences.
## Each element present in a sequences have a fixed position called as index position.

## a list can strore homogenous as well hetrogenous elements

```python
# Homogenous List : elements having similar datatypes.
x1 = ['a','b','c','d','f']
x1
```

```python
# Hetrogenous list : elements having different datatypes.
x2 = [12.34,'Apple','234',3+9j,[1,2,3]]
print(x2)
print(type(x2))
```

## Indexing :
### Accessing any specific element using its idex position.
### syntax : variable_name[index_position]

```python
print(x2[3])
print(x2[-2])
```

## Slicing :- 
### Accessing of mulitiple values at a time.
### syntax : variablename[start : stop+1:step]

```python
print(x2[2:4:1])
```

```python
print(x2[-1::-1])
print(x2[-1:-6:-1])
print(x2[::-1])
```

#### List Attributes : 

##### 1. Adding element to a list : 
<pre> 1. append(value) : add the new value to the end of the list</pre>

```python
p1 = [12.34,'Apple','234',3+9j,[1,2,3]]
p1.append('dog')
p1.append('Tiger')
print(p1)
```

<pre> 2. insert(index_position, element): add new element to a specific index position.</pre> 
<pre>syntax : variable_name.insert(first_index_value_position+1, value)  or
        variable_name.insert(second_index_value_position, value)</pre></pre>

```python
p1.insert(3,'Donkey')
print(p1)
```

<pre> 3. extend([sequence]) : add multiple elements to the end of the original list.</pre>
<pre>syntax : var_name[value]</pre>

```python
p1.extend(["Monkey","Rubi",123,"kumang"])
print(p1)
```

```python
y = [1,2,3,4,5]
z = ['a','b','c','b','c','c']
z.extend(y)
print(z)
```

```python
z.insert(2,["g",'h','i'])
z
```
