## 1. Arithmetic Operator : 
<b> Used to perform arithmetic/basic mathematical operations.</b>
<pre> +, -, *, / (single / called single division ), // (double // is called floor division), % (module), ** (powering/expontial) </pre>

<pre> Single division (/) : returns the quotient along with decimal values.
floor division(//) : returns the quotient without deciaml values that is only the interger.
Module(%) : Returns the reminder before a decimal in quotient.
Expontial/Powering(**) : Returns the result as exponent raised to the base.</pre>

```python
x = 25
y = 20
print("Addition : ", x + y)
print("Substraction : ", x-y)
print("Multiplication : ", x*y)
print("Single division : ", x/y)
print("Floor Division : ", x//y)
print("Module : " , x%y)
print("Power/Expontial : ", x**y)
```

```python
p = 134
q = 45
print(f" {p} is greater than {q} ? : ", p > q)
print(f" {p} is less than {q} ? : ", p < q)
print(f" {p} is greater than equal to {q} ? : ", p >= q)
print(f" {p} is less than equal to {q} ? : ", p <= q)
print(f" {p} is equal equal to {q} ? : ", p == q)
print(f" {p} is not equal to {q} ? : ", p != q)
print(" cat == Cat ? : " , "cat"=="Cat")
print("[1,2,3] != (1,2,3) ? : ", [1,2,3] != (1,2,3))
```
