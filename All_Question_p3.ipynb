### Q1. Create a new list where positive numbers remain unchanged and negative numbers become 0.
<pre>
    numbers = [10, -5, 20, -8, 15, -2, 30]
</pre> 

```python
new_list = [i if i > 0 else 0 for i in [10, -5, 20, -8, 15, -2, 30]]
print(new_list)
```

### Q2. Using list comprehension, create a list containing elements present in both lists.
 <pre>
list1 = [1, 2, 3, 4, 5, 6]
list2 = [4, 5, 6, 7, 8, 9]
 </pre>

```python
list1 = [1, 2, 3, 4, 5, 6]
list2 = [4, 5, 6, 7, 8, 9]
list_com = [i for i in list1 for j in list2 if i == j]
print(list_com)
```

```python
list1 = [1, 2, 3, 4, 5, 6]
list2 = [4, 5, 6, 7, 8, 9]
list_com = [i for i in list1 if i in list2]
print(list_com)
```

```python
list1 = [1, 2, 3, 4, 5, 6]
list2 = [4, 5, 6, 7, 8, 9]
list_cot = [i for i in list1+list2]
print(list_cot)
```

### Q3. Create a list of words in uppercase having more than 4 characters.
 <pre>words = ["python", "java", "javascript", "c++", "react"]</pre>

```python
words = ["python", "java", "javascript", "c++", "react"]
up_word = [i.upper() if len(i) >= 4 else i for i in words]
print(up_word)
```

### Q4. Using set comprehension, create a set containing unique even numbers.
<pre>numbers = [10, 15, 20, 20, 25, 30, 30, 35, 40]</pre>

```python
numbers = [10, 15, 20, 20, 25, 30, 30, 35, 40]
set_cot = {i for i in numbers if i % 2 == 0}
print(set_cot)
```

### 5. Create a new dictionary containing only students who scored 70 or above.
 <pre>marks = { "Rahul": 85, "Amit": 45, "Priya": 92, "John":38, "Sneha": 76} </pre>

```python
marks = { "Rahul": 85, "Amit": 45, "Priya": 92, "John":38, "Sneha": 76}
new_dict = {i:marks[i] for i in marks if marks[i] >= 70}
print(new_dict)
```

### Q6. Create a single list containing only numbers greater than 5.
<pre>

</pre>

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
list_gre = [v for i in matrix for v in i if v >= 5]
print(list_gre)
```

### Q7. Create a dictionary using comprehension that represents each unique character and its frequency.
<pre>
Eg: ip >> ‘banana’
op >> {‘b’ : 1, ‘a’ : 3, ‘n’ : 2}
</pre>

```python
usr_inp = input("Enter a string : ")
dic_up = {i: usr_inp.count(i) for i in usr_inp}
print(dic_up)
```

### Q8. Create a single list containing words whose length is greater than 4, converted to uppercase.
<pre>
    data = [ ["python", "java", "c"], ["react", "django", "html"],["sql", "excel", "powerbi"] ]
</pre>

```python
data = [["python", "java", "c"],["react", "django", "html"],["sql", "excel", "powerbi"]]
word_data = [v.upper() for i in data for v in i if len(v) >= 4]
print(word_data)
```

### Q9. Using a list comprehension generate every possible pair, but do not include pairs where both numbers are the same.
 <pre>numbers = [1, 2, 3, 4, 5]</pre>


```python
numbers = [1, 2, 3, 4, 5]
for i in numbers:
    for j in numbers:
        if i != j:
            print(i,j, end=", ")
```

```python
numbers = [1, 2, 3, 4, 5]
num_list = [(i,j) for i in numbers for j in numbers if i != j]
print(num_list)
```

### Q10. Using a dictionary comprehension display the output where: Marks >= 50 should display "Pass" and Marks < 50 should display "Fail" along with their names.
<pre>
     marks = { "Rahul": 85, "Amit": 45, "Priya": 92, "John": 38,"Sneha": 76 }
</pre>


```python
marks = { "Rahul": 85, "Amit": 45, "Priya": 92, "John": 38,"Sneha": 76 }
dist_com = {i : "Pass" if marks[i] >= 50 else "Fail" for i in marks}
print(dist_com)
```
