## <b> List Comprehension :- </b>

<b>List Comprehension is a concise way to create a list a using a single line of code. It's also help to implements for loop in it wihout writing a block of code</b>
<pre>
    Syntax : 
    <b>varname = [statment for i in sequence]</b>
</pre>

```python
# Example: To print all the values in a range within a list.
res = []
for i in range(1,11):
    res.append(i)
res
```

```python
ans = [i for i in range(1,11)]
ans
```

### Q. Using list comprehension create a list  stroing multiples of 5 from a range of 10 to 20.

```python
mul5 = [i for i in range(10,21,5)]
mul5
```

<pre> Syntax of List Comprehension with condition : 
<b>1. varname = [expression if condition else expression for i in sequence]</b>
</pre>
<pre>
    Syntax of List Comprehension if not require "if else statments. only if"
    <b>2. varname = [expression for i in sequence if condition]</b>
</pre>

```python
new = []
for i in range(11,50):
    if i % 5 == 0:
        new.append(i)
print(new)
```

```python
mult5 = [i for i in range(11,50) if i%5 == 0]
mult5
```

### Q. using a list comprehension flatten a nested list.
<pre>[[1,2][3,4][5,6]]</pre>

```python
flt = [[1,2],[3,4],[5,6]]
ne = []
for i in flt:
    ne.extend(i)
print(ne)
```

```python
for i in [[1,2],[3,4],[5,6]]:
    for val in i:
        print(val, end=" ")
```

```python
flatten_list = [value for i in [[1,2],[3,4],[5,6]] for value in i]
print(flatten_list)
```

## Dictionary Comprehension : 
<pre>
    Syntax : 
    varname = {key : value for i in sequence}

    syntax2 : 
    varname = {key : value for i in sequence if condition}

    syntax3: 
    varname : {key : statement if condition else statment for i in sequence} 
</pre>

```python
squre = {x: x**2 for x in range(11,21)}
squre
```

```python
squre_even = {x: x**2 for x in range(11,21) if x % 2 == 0}
squre_even
```
