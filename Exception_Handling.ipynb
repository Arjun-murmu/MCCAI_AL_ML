# Exception and Exception Handling :

<pre>
    - Errors are the problems in a program due to which the execution of a program stops.
    - In Errors, there are many predefined and built-in errors which are already defined in a Python library that interrupts the normal flow of our program.

    <b>Different Types of Built-in Errors in Python Library</b>
    1. NameError :- Occurs when the variable declaration does not match with the variable used in the program.
    2. TypeError : Occurs when an operation is being done on incompatible data/operands.
    3. IndexError : Occurs when an element is accessed from any sequence using an index position that does not exist.
    4. KeyError : Occurs when a key is accessed from a dictionary that does not exist.
    5. IndentationError : Occurs when the statements are declared with improper spacing/indentation.
    6. AttributeError :
        There are two types :
            a) Type -1 : Occurs when we use a predefined function/attribute of a datatype that is not associated with it.
            b) Type -2 : Occurs when a variable's name declared inside a class does not match with the variable being used.
    7. SyntaxError : Occurs when the program is not written according to its proper grammatical rules.
    8. ModuleNotFoundError : Often occurs when a module/library is being imported using the wrong name.
    9. ValueError : Occurs when the data provided by a user does not match with the datatype declared.
    11. OverflowError : Occurs when the computed answer is very large for representation.
    12. ImportError : Occurs when a module is being imported from a library that does not exist.
    14. Logical Error : These are the errors caused due to an illogical condition given in a program. These errors cannot be handled using any techinque.
</pre>

## Error Handling : 
<pre>Error Handling is a process to capture any error occurring in a program, in order to revoke or cancel the unexprected termination of a program with more user-friendly message.</pre>
* Error Handling can be done in four ways or statements
  1. try
  2. except
  3. else
  4. finally
<pre>
<b>Try : </b>
try statement are use to catch the error present within a block of code
<b>except</b>
Except block help to handel the error with more user understandable message.
<b>else : </b>
else block content the result which to be executed if there is no error in the program.
<b>finally : </b>
this block contents the code which is executed e-respective of any condition of the program.
</pre>

```python
# code without try-except statement: 
a = int(input("Enter 1st number : "))
b = int(input("Enter 2nd number : "))
result = a/b
print(f"The answer of {a}/{b} is : {result}")
```

```python
# code with try-except statements : 
try: 
    a = int(input("Enter 1st number : "))
    b = int(input("Enter 2nd number : "))
    result = a/b
    print(f"The answer of {a}/{b} is : {result}")
except:
    print("Something is wrong in the code.")
# Ig there is no error in the code, the no except block will be executed.
```

```python
# code with try-except statements : 
try: 
    a = int(input("Enter 1st number : "))
    b = int(input("Enter 2nd number : "))
    result = a/b
    print(f"The answer of {a}/{b} is : {result}")
except:
    print("Something is wrong in the code.")
```

```python
# code with try-except statements : 
try: 
    a = int(input("Enter 1st number : "))
    b = int(input("Enter 2nd number : "))
    result = a/b
    print(f"The answer of {a}/{b} is : {result}")
except:
    print("Something is wrong in the code.")
```

```python
try: 
    a = int(input("Enter 1st number : "))
    b = int(input("Enter 2nd number : "))
    result = a/b
    print(f"The answer of {a}/{b} is : {result}")
except Exception as e: # Exception is pre defind class
    print(e)   #display the error message which is alredy defined in python libary.
    print(type(e))  #displays the types of error caused in the program.
```

```python
# try with multiple except statement
try: 
    a = int(input("Enter 1st number : "))
    b = int(input("Enter 2nd number : "))
    result = a/b
    print(f"The answer of {a}/{b} is : {result}")
except ZeroDivisionError:
    print("Cannot divide a number with denominator 0.")
except ValueError:
    print("Check the datatype declared and value provided.")
```

```python
# try with multiple except statement
try: 
    a = int(input("Enter 1st number : "))
    b = int(input("Enter 2nd number : "))
    result = a/b
    print(f"The answer of {a}/{b} is : {result}")
except ZeroDivisionError:
    print("Cannot divide a number with denominator 0.")
except ValueError:
    print("Check the datatype declared and value provided.")
```

```python
# try with multiple except and else statement
try: 
    a = int(input("Enter 1st number : "))
    b = int(input("Enter 2nd number : "))
    result = a/b
except ZeroDivisionError:
    print("Cannot divide a number with denominator 0.")
except ValueError:
    print("Check the datatype declared and value provided.")
else:
    print(f"The answer of {a}/{b} is : {result}")
```

```python
# try with multiple except, else and finally statement
try: 
    a = int(input("Enter 1st number : "))
    b = int(input("Enter 2nd number : "))
    result = a/b
except ZeroDivisionError:
    print("Cannot divide a number with denominator 0.")
except ValueError:
    print("Check the datatype declared and value provided.")
else:
    print(f"The answer of {a}/{b} is : {result}")
finally:
    print("I am the finally block.")
```
