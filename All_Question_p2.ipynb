### Q.1 Wapp to count the numbers of even and odd numbers from a series of numbers.
<pre>
   series = [11,13,87,41,51,5,16,44,90,3,23,56,78,33,98,76,99] 
</pre>

```python
series = [11,13,87,41,51,5,16,44,90,3,23,56,78,33,98,76,99]
even = []
odd = []
for i in series:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
print("Series : ",series)
print("Even Number : ",even)
print("Odd Number : ",odd)
```

### Q.3 Wapp to check if the entered string is palindrome or not.
<pre>
    input -> 'mom'
    output -> 'mom' when reversed
</pre>

```python
string = input("Enter a string : ")
r_str = ''
for i in string:
    r_str = i + r_str
if r_str == string:
    print(f"{string} String is Palindrome.")
else:
    print(f"{string} String is not Palindrome.")
```

```python
string = input("Enter a string : ")
st_r = string[::-1]
if string == st_r:
    print("Palindrome Number.")
else:
    print("Not Palindrome.")
```

### Q Create a function isArmstrong and use the function to find all the armstrong numbers present between 100~2000.

```python
def isArmstrong(start,end):
    for i in range(start,end+1):
        x = len(str(i))
        sum = 0
        temp = i
        while i > 0:
            d = i % 10
            sum = sum + (d**x)
            i = i//10
        if temp == sum:
            print(temp, end = " ")
isArmstrong(int(input("Enter a start number : ")),int(input("Enter a start number : ")))
```

### Q. Create a function to find all the perfect numbers present between the specificed range given by the user.
<pre>
    When a number is equal to the sum of all its positive divisors except itself, it is said to be a perfect number.
    * Ex: 6
    sum of divisors is 1 + 2 + 3 = 6.
    Hence, 6 is a perfect number.
</pre>

```python
def isPerfect(st,end):
    for i in range(st,end):
        sum = 0
        for j in range(1,i):
            if i % j == 0:
                sum = sum + j
        if i == sum:
            print(i,end=" ")
isPerfect(int(input("Enter start number : ")),int(input("Enter end nuber : ")))
```

### Q. Create a funtion to validate a password where the criteias are: 
<pre>1. Length of the passowrd must be >= 8.
2. must contain atleast >= 1 uppercase character.
3. must contain atleast >= 1 lowercase character.
4. must contain atleast >= 1 digit.
5. must contain atleast >= 1 special character.
6. speace is not allowed between the password.
</pre>

```python
user_input = input("Enter password : ")
if len(user_input) >= 8:
    d=special=space=upp=low=0 
    for i in user_input:
        if i.isalpha() and i.isupper():
            upp += 1
        elif i.isalpha() and i.islower():
            low += 1
        elif i.isdigit():
            d += 1
        elif i.isalnum() == False and i.isspace() == False:
            special += 1
        else:
            space += 1
    if upp >= 1 and low >= 1 and d >= 1 and special >= 1 and space < 1:
        print("Valid Password.")
    else:
        print("Invalid Password.")
    print("Digit : ",d)
    print("Special : ",special)
    print(f"space {space}, Upper : {upp} ,lower : {low}")
else:
    print("Password most be 8 digit.")
```
